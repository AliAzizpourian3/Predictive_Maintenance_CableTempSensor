### pre-process the datasets one by one!

- All_sessions

```python
# 1. Load file
sessions_path = os.path.join(DATA_DIR, "AllSessions.xlsx")

sessions_df = pd.read_excel(sessions_path)

# 2. Convert timestamp
sessions_df["@timestamp"] = pd.to_datetime(sessions_df["@timestamp"])

# 3. Standardize outlet
def clean_outlet(val):
    if isinstance(val, str) and "DC" in val:
        return val.replace("DC", "")
    return str(val).strip('"')

sessions_df["outlet"] = sessions_df["outlet"].apply(clean_outlet)

# 4. Add extreme temp flag
sessions_df["is_extreme_temp"] = (
    (sessions_df["pt1000_1_T"] < -40) | (sessions_df["pt1000_1_T"] > 150) |
    (sessions_df["pt1000_2_T"] < -40) | (sessions_df["pt1000_2_T"] > 150)
)

# 5. Add valid session flag
sessions_df["is_valid_session"] = (sessions_df["energy"] > 0) & (sessions_df["duration"] > 0)

# 6. Print checks
print("✅ All_sessions cleaned")
print(f"Rows: {len(sessions_df)}")
print("🔠 Column types:")
print(sessions_df.dtypes)
print("\n🧪 Unique outlet values:", sessions_df["outlet"].unique())
print("🟡 Extreme temp rows:", sessions_df["is_extreme_temp"].sum())
print("🟢 Valid sessions:", sessions_df["is_valid_session"].sum())
```

- Exact match! Only 42 was found so decided to try Fuzzy match.............................! 

```python
# Ensure outlet and timestamp formats are compatible
filtered_flags_df["outlet"] = filtered_flags_df["outlet"].astype(str)
filtered_events_df["outlet"] = filtered_events_df["outlet"].astype(str)

filtered_flags_df["@timestamp"] = pd.to_datetime(filtered_flags_df["@timestamp"])
filtered_events_df["@timestamp"] = pd.to_datetime(filtered_events_df["@timestamp"])

# Exact inner join on timestamp, logStream, and outlet
exact_matches_df = pd.merge(
    filtered_flags_df,
    filtered_events_df,
    on=["@timestamp", "@logStream", "outlet"],
    how="inner",
    suffixes=("_flags", "_events")
)

# Check result
print(f"✅ Exact matches found: {len(exact_matches_df)}")
display(exact_matches_df.head())
```

### Fuzzy Match
- For each row in filtered_events_df, find a matching row in filtered_flags_df if:

- @logStream matches (same charger)

- outlet matches

- @timestamp is within ±60 seconds (1-minute window)

```python
# Step 1: Convert timestamp columns to datetime (just to be sure)
filtered_flags_df["@timestamp"] = pd.to_datetime(filtered_flags_df["@timestamp"])
filtered_events_df["@timestamp"] = pd.to_datetime(filtered_events_df["@timestamp"])

# Step 2: Pre-sort both DataFrames
filtered_flags_df = filtered_flags_df.sort_values("@timestamp")
filtered_events_df = filtered_events_df.sort_values("@timestamp")

# Step 3: Set time window for matching
time_window = timedelta(seconds=60)

# Step 4: Prepare a list to store matched flag rows
matched_flag_rows = []

# Step 5: Fuzzy match loop
for _, event_row in filtered_events_df.iterrows():
    event_time = event_row["@timestamp"]
    logstream = event_row["@logStream"]
    outlet = event_row["outlet"]

    # Get subset of flags with same charger and outlet
    candidate_flags = filtered_flags_df[
        (filtered_flags_df["@logStream"] == logstream) &
        (filtered_flags_df["outlet"] == outlet) &
        (filtered_flags_df["@timestamp"].between(event_time - time_window, event_time + time_window))
    ]

    # Take the nearest flag row (if available)
    if not candidate_flags.empty:
        nearest_flag = candidate_flags.iloc[(candidate_flags["@timestamp"] - event_time).abs().argsort()].iloc[0]
        combined_row = {**event_row.to_dict(), **nearest_flag.to_dict()}
        matched_flag_rows.append(combined_row)
# Step 6: Combine matched rows into one DataFrame
fuzzy_matches_df = pd.DataFrame(matched_flag_rows)

print(f"✅ Matched rows found: {len(fuzzy_matches_df)}")
display(fuzzy_matches_df.head())
```

### summary of mean Tempdiff for each flag

```python
# Create a summary of mean Tempdiff for each flag
flag_columns = ['RNE', 'PLE', 'SOV', 'COV', 'CBL', 'KON', 'HOU', 'CFeed', 'CRet']

summary_stats = {}

for flag in flag_columns:
    with_flag = fuzzy_matches_df[fuzzy_matches_df[flag] == 1]['Tempdiff']
    without_flag = fuzzy_matches_df[fuzzy_matches_df[flag] != 1]['Tempdiff']
    
    summary_stats[flag] = {
        'mean_with_flag': with_flag.mean(),
        'mean_without_flag': without_flag.mean(),
        'count_with_flag': with_flag.count(),
        'count_without_flag': without_flag.count()
    }

summary_df = pd.DataFrame(summary_stats).T
display(summary_df.sort_values("mean_with_flag", ascending=False))
```

#### Summary based on typw of flag and temp!

```python
# Focus on key columns
cols_to_analyze = ['type', 'Tempdiff'] + flag_columns  # flag_columns was defined earlier

# Group by 'type' and compute mean values
grouped_by_type = fuzzy_matches_df[cols_to_analyze].groupby('type').mean().T

# Count of flags triggered by type
flag_counts = fuzzy_matches_df.groupby('type')[flag_columns].sum().T

# Combine both
combined_flag_analysis = pd.concat([grouped_by_type, flag_counts], axis=1, keys=['Mean_Values', 'Total_Counts'])

# Display
display(combined_flag_analysis)
```

#### Clustering

```python
# Load the combined matched dataset (assuming it's already loaded or passed)
file_path = "../data/Matched_Flagged_Events.xlsx"
df = pd.read_excel(file_path)

# Select relevant features for clustering
features = ['Tempdiff', 'PLE', 'RNE', 'KON']
clustering_data = df[features].fillna(0)

# Normalize the features
scaler = StandardScaler()
scaled_features = scaler.fit_transform(clustering_data)

# Apply KMeans clustering
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(scaled_features)

# Add event type to see distribution
if 'type' in df.columns:
    cluster_summary = df.groupby(['cluster', 'type']).size().unstack(fill_value=0)
else:
    cluster_summary = df['cluster'].value_counts().to_frame('count')

# Plot the cluster assignments using Tempdiff vs PLE
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Tempdiff', y='PLE', hue='cluster', palette='Set2', s=70)
plt.title("Clustering based on Tempdiff and Flags")
plt.xlabel("Temperature Difference")
plt.ylabel("PLE Flag")
plt.grid(True)
plt.tight_layout()
plt.show()

# Show the cluster summary
print(cluster_summary)
```
